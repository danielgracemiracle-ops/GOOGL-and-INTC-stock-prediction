# 3A - Generative Adversarial Network (GAN)

## Cara Kerja Generative Adversarial Network (GAN)

Bayangkan jaringan yang belajar menciptakan gambar hingga terlihat nyata. Itulah inti dari Generative Adversarial Network, atau GAN. Diciptakan oleh Ian Goodfellow di tahun 2014, sistem ini mulai menarik perhatian ilmuwan. Cara kerjanya unik: dua bagian saling beradu, satu membuat tiruan, satunya lagi menilai keaslian. Hasilnya? Data buatan yang sangat mirip dengan contoh aslinya. Teknologi semacam ini kemudian membentuk dasar bagi banyak inovasi dalam bidang AI generatif. Tidak heran jika penggunaannya kini makin meluas di berbagai sektor.

Arsitektur GAN terdiri dari dua komponen utama, yaitu:

### 1. gambar

Gambar punya tugas mencipta citra buatan dari angka-angka tak beraturan. Dalam kekacauan itulah bentuk mulai muncul perlahan. Titik awalnya hanyalah deretan nilai yang tampak tanpa makna. Namun lambat-laun, struktur mengambil alih. Proses ini membangun visual mirip contoh nyata dalam kumpulan data. Hasil akhir bisa sangat menyerupai foto atau lukisan asli.

Bukan soal menjiplak, tapi mendekati tampilan asli sampai pembatasnya kabur. Dari sudut pandang mesin, hasil ini harus terasa biasa - seolah tidak ada rekayasa sama sekali. Bayangkan saja: satu perbedaan kecil bisa membuat semua berubah. Meski begitu, fokus utamanya tetap pada ketidakkelihatan - seperti bayangan yang tak ingin dikenali.

### 2. Discriminator

Pada dasarnya, tugas utama discriminator adalah membedakan - apakah sebuah citra berasal dari kumpulan data nyata atau justru dibuat oleh sistem. Terkadang ia bekerja seperti penapis, mengamati pola-pola halus yang muncul dalam visual. Gambar real akan tampak lebih alami baginya, sementara yang palsu sering meninggalkan jejak aneh di tekstur. Ia tidak pernah langsung percaya pada apa yang dilihat. Setiap masukan dinilai berdasarkan pengalaman sebelumnya terhadap ribuan contoh lain.

Output Discriminator berupa probabilitas yang menunjukkan tingkat keyakinan bahwa suatu gambar merupakan gambar asli.

Pembuat keputusan ini harus bisa mengenali mana citra benaran, mana yang cuma tiruan - dengan ketepatan sebesar mungkin. Mulai dari sana, tugasnya jelas: menangkap perbedaan halus tanpa gagal. Tidak ada ruang untuk ragu saat mata melihat sesuatu yang tidak wajar. Begitu sistem bekerja, setiap detil ikut bicara. Terkadang yang kelihatan normal malah mencurigakan. Hasil akhir? Pemisahan antara nyata dan buatan terjadi begitu saja.

---

## Mekanisme Training GAN

Proses pelatihan GAN berlangsung secara adversarial (saling berlawanan) antara gambar dan Discriminator.

### Langkah 1: gambar Membuat gambar Palsu

gambar menerima input berupa random noise kemudian menghasilkan fake image.

### Langkah 2: Discriminator Menerima Dua Jenis Data

Discriminator menerima:

* Real image dari training set
* Fake image dari gambar

### Langkah 3: Discriminator Melakukan Klasifikasi

Discriminator memprediksi apakah gambar yang diterima merupakan:

* Real
* Fake

Kemudian loss Discriminator dihitung berdasarkan hasil prediksi tersebut.

### Langkah 4: gambar Diperbarui

gambar diperbarui menggunakan feedback dari Discriminator. gambar berusaha menghasilkan gambar yang semakin realistis agar dapat "menipu" Discriminator.

### Langkah 5: Proses Berulang

gambar dan Discriminator dilatih secara bergantian hingga mencapai kondisi keseimbangan (equilibrium), yaitu ketika Discriminator tidak lagi dapat membedakan secara jelas antara gambar asli dan gambar hasil generasi.

---

## Ilustrasi Berdasarkan Gambar

Pada gambar, random noise digunakan sebagai input ke gambar untuk menghasilkan fake image. Selanjutnya, fake image dan real image dari training set diberikan ke Discriminator. Discriminator kemudian menentukan apakah gambar tersebut termasuk kategori real atau fake. Selama proses training, gambar berusaha menghasilkan gambar yang semakin realistis, sedangkan Discriminator berusaha meningkatkan kemampuannya dalam membedakan gambar asli dan gambar palsu.

---

## Kesimpulan

GAN merupakan arsitektur deep learning yang terdiri dari gambar dan Discriminator yang dilatih secara kompetitif. gambar bertujuan menghasilkan gambar yang menyerupai data asli, sedangkan Discriminator bertujuan membedakan gambar asli dan gambar hasil generasi. Melalui proses adversarial training, GAN mampu menghasilkan gambar sintetis yang memiliki distribusi yang semakin mendekati distribusi data asli.


# 3B - Baseline GAN for Image Generation

## Tujuan

Pada bagian ini akan dibangun arsitektur Generative Adversarial Network (GAN) untuk melakukan image generation menggunakan dataset Overhead-MNIST kelas Stadium. Tujuan utama dari model ini adalah mempelajari distribusi gambar Stadium sehingga mampu menghasilkan gambar sintetis yang memiliki karakteristik serupa dengan gambar asli.

GAN terdiri dari dua komponen utama, yaitu Generator dan Discriminator. Generator bertugas menghasilkan gambar sintetis dari random noise, sedangkan Discriminator bertugas membedakan gambar asli dan gambar hasil generasi.

## Arsitektur Baseline GAN

Karena data yang digunakan berupa gambar berukuran 28×28 piksel, arsitektur baseline menggunakan pendekatan Deep Convolutional Generative Adversarial Network (DCGAN). Penggunaan layer konvolusional memungkinkan model mempelajari pola spasial gambar dengan lebih baik dibandingkan layer fully connected biasa.

### Generator

Generator menerima input berupa random noise vector berdimensi 100 dan mengubahnya menjadi gambar Stadium berukuran 28×28 piksel menggunakan beberapa layer transposed convolution.

Struktur Generator:

* Input Noise Vector (100 dimensi)
* Fully Connected Layer
* Reshape Feature Map
* ConvTranspose Layer
* Batch Normalization
* ReLU Activation
* ConvTranspose Layer
* Tanh Activation
* Output Image (28×28)

Generator bertujuan menghasilkan gambar sintetis yang semirip mungkin dengan gambar Stadium asli sehingga sulit dibedakan oleh Discriminator.

### Discriminator

Discriminator menerima gambar asli maupun gambar hasil generasi kemudian menentukan apakah gambar tersebut termasuk kategori Real atau Fake.

Struktur Discriminator:

* Input Image (28×28)
* Convolution Layer
* LeakyReLU Activation
* Convolution Layer
* LeakyReLU Activation
* Fully Connected Layer
* Sigmoid Activation
* Output Probability (Real/Fake)

Discriminator berfungsi sebagai classifier biner yang memberikan umpan balik kepada Generator selama proses training.

## Hyperparameter Baseline

| Parameter        | Nilai                      |
| ---------------- | -------------------------- |
| Latent Dimension | 100                        |
| Batch Size       | 64                         |
| Epoch            | 50                         |
| Learning Rate    | 0.0002                     |
| Optimizer        | Adam                       |
| Loss Function    | Binary Cross Entropy (BCE) |

## Evaluasi

Kualitas gambar yang dihasilkan akan dievaluasi menggunakan Fréchet Inception Distance (FID). Metrik ini membandingkan distribusi feature antara gambar hasil generasi dan gambar asli menggunakan representasi feature dari jaringan Inception. Semakin kecil nilai FID, semakin mirip distribusi gambar hasil generasi terhadap distribusi gambar asli, yang menunjukkan kualitas image generation yang lebih baik.


```python
# import library
import os
import numpy as np
import matplotlib.pyplot as plt
import glob

from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader, random_split
from torchvision import transforms

from tqdm import tqdm
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
```

```python
# Load Gambar Stadium
stadium_train = glob.glob(
    "version2/train/stadium/*.jpg"
)

stadium_test = glob.glob(
    "version2/test/stadium/*.jpg"
)

all_stadium = stadium_train + stadium_test

print("Total Images :", len(all_stadium))
```

```python
# transform
from torchvision import transforms

transform = transforms.Compose([
    transforms.Grayscale(),
    transforms.Resize((28,28)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])
```

```python
# custom dataset
from PIL import Image
from torch.utils.data import Dataset

class StadiumDataset(Dataset):

    def __init__(self, image_paths, transform=None):

        self.image_paths = image_paths
        self.transform = transform

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):

        image = Image.open(
            self.image_paths[idx]
        ).convert("L")

        if self.transform:
            image = self.transform(image)

        return image
```

```python
# dataset object
dataset = StadiumDataset(
    all_stadium,
    transform=transform
)

print("Dataset Size :", len(dataset))
```

```python
# split 80-10-10
from torch.utils.data import random_split

total_size = len(dataset)

train_size = int(0.8 * total_size)
val_size = int(0.1 * total_size)

test_size = (
    total_size
    - train_size
    - val_size
)

train_dataset, val_dataset, test_dataset = random_split(
    dataset,
    [train_size, val_size, test_size]
)

print("Train :", len(train_dataset))
print("Val   :", len(val_dataset))
print("Test  :", len(test_dataset))
```

```python
# Data Loader
from torch.utils.data import DataLoader

batch_size = 64

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False
)
```

```python
# visualisasi sample
import matplotlib.pyplot as plt

sample_batch = next(
    iter(train_loader)
)

plt.figure(figsize=(8,4))

for i in range(8):

    plt.subplot(2,4,i+1)

    img = sample_batch[i].squeeze()

    plt.imshow(
        img,
        cmap="gray"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()
```

## Data Preparation

Dataset Stadium berhasil dimuat dan dilakukan preprocessing berupa konversi grayscale, resizing menjadi 28×28 piksel, normalisasi ke rentang [-1,1], serta pembagian data menjadi training set, validation set, dan test set. Data training digunakan untuk melatih GAN, sedangkan data test digunakan untuk evaluasi menggunakan Fréchet Inception Distance (FID).

```python
# Generator (Baseline DCGAN)
import torch
import torch.nn as nn

class Generator(nn.Module):

    def __init__(self, latent_dim=100):

        super().__init__()

        self.model = nn.Sequential(

            nn.Linear(latent_dim, 128 * 7 * 7),

            nn.ReLU(True)
        )

        self.conv = nn.Sequential(

            nn.ConvTranspose2d(
                128,
                64,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.BatchNorm2d(64),

            nn.ReLU(True),

            nn.ConvTranspose2d(
                64,
                1,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.Tanh()
        )

    def forward(self, z):

        x = self.model(z)

        x = x.view(-1, 128, 7, 7)

        x = self.conv(x)

        return x
```

```python
# Discriminator (Baseline DCGAN)
class Discriminator(nn.Module):
    
    def __init__(self):

        super().__init__()

        self.model = nn.Sequential(

            nn.Conv2d(
                1,
                64,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.LeakyReLU(
                0.2,
                inplace=True
            ),

            nn.Conv2d(
                64,
                128,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.LeakyReLU(
                0.2,
                inplace=True
            ),

            nn.Flatten(),

            nn.Linear(
                128 * 7 * 7,
                1
            ),

            nn.Sigmoid()
        )

    def forward(self, x):

        return self.model(x)
```

```python
# Inisialisasi Model
device = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

latent_dim = 100

generator = Generator(
    latent_dim
).to(device)

discriminator = Discriminator().to(device)

print(generator)
print()
print(discriminator)
```

```python
# Lost Function and Optimizer
criterion = nn.BCELoss()

optimizer_G = torch.optim.Adam(
    generator.parameters(),
    lr=0.0002,
    betas=(0.5, 0.999)
)

optimizer_D = torch.optim.Adam(
    discriminator.parameters(),
    lr=0.0002,
    betas=(0.5, 0.999)
)
```

## Training Baseline GAN

Model GAN dilatih menggunakan pendekatan adversarial learning. Pada setiap iterasi, Discriminator terlebih dahulu dilatih untuk membedakan gambar asli dan gambar hasil generasi. Selanjutnya Generator dilatih untuk menghasilkan gambar yang mampu menipu Discriminator.

Loss Function yang digunakan adalah Binary Cross Entropy (BCE), sedangkan optimizer yang digunakan adalah Adam dengan learning rate sebesar 0.0002.

```python
# Training Loop
epochs = 50

g_losses = []
d_losses = []

for epoch in range(epochs):

    g_loss_epoch = 0
    d_loss_epoch = 0

    for real_imgs in train_loader:

        real_imgs = real_imgs.to(device)

        batch_size = real_imgs.size(0)

        real_labels = torch.ones(
            batch_size,
            1
        ).to(device)

        fake_labels = torch.zeros(
            batch_size,
            1
        ).to(device)

        # ==================
        # Train Discriminator
        # ==================

        optimizer_D.zero_grad()

        outputs_real = discriminator(
            real_imgs
        )

        loss_real = criterion(
            outputs_real,
            real_labels
        )

        z = torch.randn(
            batch_size,
            latent_dim
        ).to(device)

        fake_imgs = generator(z)

        outputs_fake = discriminator(
            fake_imgs.detach()
        )

        loss_fake = criterion(
            outputs_fake,
            fake_labels
        )

        d_loss = (
            loss_real
            + loss_fake
        ) / 2

        d_loss.backward()

        optimizer_D.step()

        # ==================
        # Train Generator
        # ==================

        optimizer_G.zero_grad()

        z = torch.randn(
            batch_size,
            latent_dim
        ).to(device)

        fake_imgs = generator(z)

        outputs = discriminator(
            fake_imgs
        )

        g_loss = criterion(
            outputs,
            real_labels
        )

        g_loss.backward()

        optimizer_G.step()

        g_loss_epoch += g_loss.item()
        d_loss_epoch += d_loss.item()

    g_losses.append(
        g_loss_epoch / len(train_loader)
    )

    d_losses.append(
        d_loss_epoch / len(train_loader)
    )

    print(
        f"Epoch [{epoch+1}/{epochs}] "
        f"D Loss: {d_losses[-1]:.4f} "
        f"G Loss: {g_losses[-1]:.4f}"
    )
```

```python
# visualisasi training loss

plt.figure(figsize=(8,5))

plt.plot(
    g_losses,
    label="Generator Loss"
)

plt.plot(
    d_losses,
    label="Discriminator Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("GAN Training Loss")

plt.legend()

plt.show()
```

## Training Loss Analysis

Grafik di atas menunjukkan perkembangan Generator Loss dan Discriminator Loss selama proses training GAN baseline.

Pada awal training (epoch 1–10), Generator Loss mengalami fluktuasi yang cukup besar. Hal ini menunjukkan bahwa Generator masih berada pada tahap awal pembelajaran untuk menghasilkan gambar yang menyerupai gambar Stadium asli. Pada saat yang sama, Discriminator mampu membedakan gambar asli dan gambar hasil generasi dengan relatif baik sehingga nilai Discriminator Loss tetap rendah.

Pada rentang epoch 10–28, kedua loss cenderung lebih stabil. Kondisi ini menunjukkan bahwa Generator dan Discriminator mulai mencapai keseimbangan sehingga proses adversarial learning berlangsung secara lebih efektif.

Namun, pada sekitar epoch 29–41 terjadi peningkatan Generator Loss yang sangat signifikan disertai penurunan Discriminator Loss mendekati nol. Fenomena ini mengindikasikan bahwa Discriminator menjadi terlalu dominan sehingga Generator mengalami kesulitan untuk menghasilkan gambar yang mampu menipu Discriminator. Ketidakseimbangan seperti ini merupakan fenomena yang umum terjadi pada proses training GAN.

Menjelang akhir training (epoch 42–50), Generator Loss kembali menurun dan kedua loss kembali berada pada rentang yang lebih stabil. Hal ini menunjukkan bahwa Generator mulai menyesuaikan parameter modelnya untuk menghasilkan gambar yang lebih realistis.

Secara keseluruhan, hasil training menunjukkan bahwa proses adversarial learning berhasil berlangsung meskipun terdapat periode ketidakseimbangan antara Generator dan Discriminator. Oleh karena itu, kualitas akhir model akan dievaluasi lebih lanjut menggunakan Fréchet Inception Distance (FID), yang membandingkan distribusi feature antara gambar hasil generasi dan gambar asli.

# 3C - Improved GAN Architecture and Hyperparameter Tuning

## Motivation for Improvement

Hasil evaluasi pada bagian 3B menunjukkan bahwa model baseline GAN memperoleh nilai:

**Baseline FID = 183.4631**

Nilai FID yang masih relatif tinggi menunjukkan bahwa distribusi gambar hasil generasi masih memiliki perbedaan yang cukup besar dibandingkan distribusi gambar asli Stadium. Selain itu, grafik training loss menunjukkan adanya periode ketidakseimbangan antara Generator dan Discriminator selama proses training. Oleh karena itu, diperlukan modifikasi arsitektur dan tuning hyperparameter untuk meningkatkan kualitas image generation.

---

## Architecture Improvements

Beberapa modifikasi dilakukan pada arsitektur baseline GAN sebagai berikut:

### 1. Batch Normalization in Generator

Batch Normalization ditambahkan pada Generator untuk menstabilkan distribusi aktivasi pada setiap layer selama proses training.

Keuntungan:

* Mempercepat proses konvergensi.
* Mengurangi internal covariate shift.
* Membantu Generator menghasilkan gambar yang lebih konsisten.
* Mengurangi kemungkinan training menjadi tidak stabil.

### 2. Dropout in Discriminator

Dropout ditambahkan pada Discriminator untuk mengurangi risiko overfitting.

Keuntungan:

* Mencegah Discriminator menjadi terlalu kuat dibanding Generator.
* Memaksa Discriminator mempelajari fitur yang lebih umum.
* Membantu menjaga keseimbangan proses adversarial learning.

### 3. Improved Weight Initialization

Metode weight initialization khusus untuk GAN digunakan agar parameter model berada pada rentang yang lebih stabil pada awal training.

Keuntungan:

* Mempercepat proses pembelajaran.
* Mengurangi risiko exploding atau vanishing gradients.
* Membantu stabilitas training GAN.

---

## Hyperparameter Tuning

Selain modifikasi arsitektur, dilakukan tuning hyperparameter sebagai berikut:

| Parameter        | Baseline | Improved |
| ---------------- | -------- | -------- |
| Epoch            | 50       | 100      |
| Learning Rate    | 0.0002   | 0.0001   |
| Batch Size       | 64       | 64       |
| Optimizer        | Adam     | Adam     |
| Latent Dimension | 100      | 100      |

### Reasoning

#### Increasing Number of Epochs

Jumlah epoch ditingkatkan dari 50 menjadi 100 agar Generator memiliki lebih banyak kesempatan untuk mempelajari distribusi gambar Stadium secara lebih mendalam.

#### Lower Learning Rate

Learning rate diturunkan dari 0.0002 menjadi 0.0001 untuk menghasilkan proses update parameter yang lebih stabil serta mengurangi fluktuasi yang berlebihan selama training GAN.

---

## Expected Outcome

Dengan penambahan Batch Normalization, Dropout, weight initialization yang lebih baik, serta tuning hyperparameter, model diharapkan mampu menghasilkan gambar sintetis yang lebih realistis dan memiliki distribusi feature yang lebih mendekati gambar asli.

Keberhasilan modifikasi akan dievaluasi menggunakan Fréchet Inception Distance (FID). Model improved dianggap lebih baik apabila menghasilkan nilai FID yang lebih rendah dibandingkan baseline GAN.


```python
# improved Generator
class ImprovedGenerator(nn.Module):
    
    def __init__(self, latent_dim=100):

        super().__init__()

        self.fc = nn.Sequential(

            nn.Linear(
                latent_dim,
                256 * 7 * 7
            ),

            nn.BatchNorm1d(
                256 * 7 * 7
            ),

            nn.ReLU(True)
        )

        self.conv = nn.Sequential(

            nn.ConvTranspose2d(
                256,
                128,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.BatchNorm2d(128),

            nn.ReLU(True),

            nn.ConvTranspose2d(
                128,
                64,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.BatchNorm2d(64),

            nn.ReLU(True),

            nn.Conv2d(
                64,
                1,
                kernel_size=3,
                stride=1,
                padding=1
            ),

            nn.Tanh()
        )

    def forward(self, z):

        x = self.fc(z)

        x = x.view(
            -1,
            256,
            7,
            7
        )

        x = self.conv(x)

        return x
```

```python
# Improved discriminator
class ImprovedDiscriminator(nn.Module):
    
    def __init__(self):

        super().__init__()

        self.model = nn.Sequential(

            nn.Conv2d(
                1,
                64,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.LeakyReLU(
                0.2,
                inplace=True
            ),

            nn.Dropout2d(
                0.3
            ),

            nn.Conv2d(
                64,
                128,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.BatchNorm2d(
                128
            ),

            nn.LeakyReLU(
                0.2,
                inplace=True
            ),

            nn.Dropout2d(
                0.3
            ),

            nn.Flatten(),

            nn.Linear(
                128 * 7 * 7,
                1
            ),

            nn.Sigmoid()
        )

    def forward(self, x):

        return self.model(x)
```

```python
# DCGAN weight initialization
def weights_init(m):
    
    classname = m.__class__.__name__

    if classname.find('Conv') != -1:

        nn.init.normal_(
            m.weight.data,
            0.0,
            0.02
        )

    elif classname.find('BatchNorm') != -1:

        nn.init.normal_(
            m.weight.data,
            1.0,
            0.02
        )

        nn.init.constant_(
            m.bias.data,
            0
        )
```

```python
# initialize Improved Model
latent_dim = 100

generator_imp = ImprovedGenerator(
    latent_dim
).to(device)

discriminator_imp = ImprovedDiscriminator().to(device)

generator_imp.apply(
    weights_init
)

discriminator_imp.apply(
    weights_init
)
```

```python
# improved Optimizer
criterion = nn.BCELoss()

optimizer_G_imp = torch.optim.Adam(
    generator_imp.parameters(),
    lr=0.0001,
    betas=(0.5, 0.999)
)

optimizer_D_imp = torch.optim.Adam(
    discriminator_imp.parameters(),
    lr=0.0001,
    betas=(0.5, 0.999)
)
```

## Training Improved GAN

Pada tahap ini dilakukan pelatihan Improved GAN menggunakan arsitektur yang telah dimodifikasi. Perubahan utama meliputi penambahan Batch Normalization pada Generator, penambahan Dropout pada Discriminator, penggunaan DCGAN Weight Initialization, peningkatan jumlah epoch, serta penyesuaian learning rate.

Tujuan dari modifikasi ini adalah meningkatkan stabilitas training, mengurangi overfitting pada Discriminator, dan menghasilkan gambar sintetis yang lebih realistis sehingga nilai Fréchet Inception Distance (FID) dapat diturunkan dibandingkan model baseline.

```python
# Training loop
epochs = 100

g_losses_imp = []
d_losses_imp = []

for epoch in range(epochs):

    g_loss_epoch = 0
    d_loss_epoch = 0

    for real_imgs in train_loader:

        real_imgs = real_imgs.to(device)

        batch_size = real_imgs.size(0)

        real_labels = torch.ones(
            batch_size,
            1
        ).to(device)

        fake_labels = torch.zeros(
            batch_size,
            1
        ).to(device)

        # ==========================
        # Train Discriminator
        # ==========================

        optimizer_D_imp.zero_grad()

        outputs_real = discriminator_imp(
            real_imgs
        )

        loss_real = criterion(
            outputs_real,
            real_labels
        )

        z = torch.randn(
            batch_size,
            latent_dim
        ).to(device)

        fake_imgs = generator_imp(z)

        outputs_fake = discriminator_imp(
            fake_imgs.detach()
        )

        loss_fake = criterion(
            outputs_fake,
            fake_labels
        )

        d_loss = (
            loss_real
            + loss_fake
        ) / 2

        d_loss.backward()

        optimizer_D_imp.step()

        # ==========================
        # Train Generator
        # ==========================

        optimizer_G_imp.zero_grad()

        z = torch.randn(
            batch_size,
            latent_dim
        ).to(device)

        fake_imgs = generator_imp(z)

        outputs = discriminator_imp(
            fake_imgs
        )

        g_loss = criterion(
            outputs,
            real_labels
        )

        g_loss.backward()

        optimizer_G_imp.step()

        g_loss_epoch += g_loss.item()
        d_loss_epoch += d_loss.item()

    avg_g_loss = (
        g_loss_epoch
        / len(train_loader)
    )

    avg_d_loss = (
        d_loss_epoch
        / len(train_loader)
    )

    g_losses_imp.append(
        avg_g_loss
    )

    d_losses_imp.append(
        avg_d_loss
    )

    print(
        f"Epoch [{epoch+1}/{epochs}] "
        f"D Loss: {avg_d_loss:.4f} "
        f"G Loss: {avg_g_loss:.4f}"
    )
```

```python
# plot loss
plt.figure(figsize=(8,5))

plt.plot(
    g_losses_imp,
    label="Improved Generator Loss"
)

plt.plot(
    d_losses_imp,
    label="Improved Discriminator Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Improved GAN Training Loss")

plt.legend()

plt.show()
```

## Improved GAN Training Loss Analysis

Grafik di atas menunjukkan perkembangan Improved Generator Loss dan Improved Discriminator Loss selama proses training Improved GAN pada dataset Stadium.

Pada awal training (epoch 1–5), Generator Loss mengalami fluktuasi yang cukup besar sebelum turun secara signifikan menuju nilai yang lebih stabil. Kondisi ini menunjukkan bahwa Generator sedang menyesuaikan parameter awal untuk mempelajari distribusi gambar Stadium. Pada rentang epoch 5–45, kedua loss menunjukkan pola yang relatif stabil dengan nilai berada di sekitar 0.7. Kondisi ini mengindikasikan bahwa Generator dan Discriminator berada pada keseimbangan yang baik sehingga proses adversarial learning berlangsung secara efektif.

Terdapat peningkatan Generator Loss pada sekitar epoch 47–55 yang diikuti oleh sedikit penurunan Discriminator Loss. Fenomena ini menunjukkan bahwa Discriminator untuk sementara memiliki kemampuan yang lebih baik dalam membedakan gambar asli dan gambar hasil generasi. Hal ini menunjukkan bahwa modifikasi arsitektur dan tuning hyperparameter yang dilakukan berhasil meningkatkan stabilitas proses training.

Secara keseluruhan, Improved GAN menunjukkan proses adversarial learning yang lebih seimbang dibandingkan baseline GAN.
