# 3A - Generative Adversarial Network (GAN)

## Cara Kerja Generative Adversarial Network (GAN)

Bayangkan jaringan yang belajar menciptakan gambar hingga terlihat nyata. Itulah inti dari Generative Adversarial Network, atau GAN. Diciptakan oleh Ian Goodfellow di tahun 2014, sistem ini mulai menarik perhatian ilmuwan. Cara kerjanya unik: dua bagian saling beradu, satu membuat tiruan, satunya lagi menilai keaslian. Hasilnya? Data buatan yang sangat mirip dengan contoh aslinya. Teknologi semacam ini kemudian membentuk dasar bagi banyak inovasi dalam bidang AI generatif. Tidak heran jika penggunaannya kini makin meluas di berbagai sektor.

Arsitektur GAN terdiri dari dua komponen utama, yaitu:

### 1. gambar

Gambar punya tugas mencipta citra buatan dari angka-angka tak beraturan. Dalam kekacauan itulah bentuk mulai muncul perlahan. Titik awalnya hanyalah deretan nilai yang tampak tanpa makna. Namun lambat-laun, struktur mengambil alih. Proses ini membangun visual mirip contoh nyata dalam kumpulan data. Hasil akhir bisa sangat menyerupai foto atau lukisan asli.

Bukan soal menjiplak, tapi mendekati tampilan asli sampai pembatasnya kabur. Dari sudut pandang mesin, hasil ini harus terasa biasa - seolah tidak ada rekayasa sama sekali. Bayangkan saja: satu perbedaan kecil bisa membuat semua berubah. Meski begitu, fokus utamanya tetap pada ketidakkelihatan - seperti bayangan yang tak ingin dikenali.

### 2. Discriminator

Pada dasarnya, tugas utama discriminator adalah membedakan - apakah sebuah citra berasal dari kumpulan data nyata atau justru dibuat oleh sistem. Terkadang ia bekerja seperti penapis, mengamati pola-pola halus yang muncul dalam visual. Gambar real akan tampak lebih alami baginya, sementara yang palsu sering meninggalkan jejak aneh di tekstur. Ia tidak pernah langsung percaya pada apa yang dilihat. Setiap masukan dinilai berdasarkan pengalaman sebelumnya terhadap ribuan contoh lain.

Output Discriminator berupa probabilitas yang menunjukkan tingkat keyakinan bahwa suatu gambar merupakan gambar asli.

Pembuat keputusan ini harus bisa mengenali mana citra benaran, mana yang cuma tiruan - dengan ketepatan sebesar mungkin. Mulai dari sana, tugasnya jelas: menangkap perbedaan halus tanpa gagal. Tidak ada ruang untuk ragu saat mata melihat sesuatu yang tidak wajar. Begitu sistem bekerja, setiap detil ikut bicara. Terkadang yang kelihatan normal malah mencurigakan. Hasil akhir? Pemisahan antara nyata dan buatan terjadi begitu saja.

---

## Mekanisme Training GAN

Proses pelatihan GAN berlangsung secara adversarial (saling berlawanan) antara gambar dan Discriminator.

### Langkah 1: gambar Membuat gambar Palsu

gambar menerima input berupa random noise kemudian menghasilkan fake image.

### Langkah 2: Discriminator Menerima Dua Jenis Data

Discriminator menerima:

* Real image dari training set
* Fake image dari gambar

### Langkah 3: Discriminator Melakukan Klasifikasi

Discriminator memprediksi apakah gambar yang diterima merupakan:

* Real
* Fake

Kemudian loss Discriminator dihitung berdasarkan hasil prediksi tersebut.

### Langkah 4: gambar Diperbarui

gambar diperbarui menggunakan feedback dari Discriminator. gambar berusaha menghasilkan gambar yang semakin realistis agar dapat "menipu" Discriminator.

### Langkah 5: Proses Berulang

gambar dan Discriminator dilatih secara bergantian hingga mencapai kondisi keseimbangan (equilibrium), yaitu ketika Discriminator tidak lagi dapat membedakan secara jelas antara gambar asli dan gambar hasil generasi.

---

## Ilustrasi Berdasarkan Gambar

Pada gambar, random noise digunakan sebagai input ke gambar untuk menghasilkan fake image. Selanjutnya, fake image dan real image dari training set diberikan ke Discriminator. Discriminator kemudian menentukan apakah gambar tersebut termasuk kategori real atau fake. Selama proses training, gambar berusaha menghasilkan gambar yang semakin realistis, sedangkan Discriminator berusaha meningkatkan kemampuannya dalam membedakan gambar asli dan gambar palsu.

---

## Kesimpulan

GAN merupakan arsitektur deep learning yang terdiri dari gambar dan Discriminator yang dilatih secara kompetitif. gambar bertujuan menghasilkan gambar yang menyerupai data asli, sedangkan Discriminator bertujuan membedakan gambar asli dan gambar hasil generasi. Melalui proses adversarial training, GAN mampu menghasilkan gambar sintetis yang memiliki distribusi yang semakin mendekati distribusi data asli.


# 3B - Baseline GAN for Image Generation

## Tujuan

Pada bagian ini akan dibangun arsitektur Generative Adversarial Network (GAN) untuk melakukan image generation menggunakan dataset Overhead-MNIST kelas Oil Gas Field. Tujuan utama dari model ini adalah mempelajari distribusi gambar Oil Gas Field sehingga mampu menghasilkan gambar sintetis yang memiliki karakteristik serupa dengan gambar asli.

GAN terdiri dari dua komponen utama, yaitu gambar dan Discriminator. gambar bertugas menghasilkan gambar sintetis dari random noise, sedangkan Discriminator bertugas membedakan gambar asli dan gambar hasil generasi.

## Arsitektur Baseline GAN

Karena data yang digunakan berupa gambar berukuran 28×28 piksel, arsitektur baseline menggunakan pendekatan Deep Convolutional Generative Adversarial Network (DCGAN). Penggunaan layer konvolusional memungkinkan model mempelajari pola spasial gambar dengan lebih baik dibandingkan layer fully connected biasa.

### gambar

gambar menerima input berupa random noise vector berdimensi 100 dan mengubahnya menjadi gambar Oil Gas Field berukuran 28×28 piksel menggunakan beberapa layer transposed convolution.

Struktur gambar:

* Input Noise Vector (100 dimensi)
* Fully Connected Layer
* Reshape Feature Map
* ConvTranspose Layer
* Batch Normalization
* ReLU Activation
* ConvTranspose Layer
* Tanh Activation
* Output Image (28×28)

gambar bertujuan menghasilkan gambar sintetis yang semirip mungkin dengan gambar Oil Gas Field asli sehingga sulit dibedakan oleh Discriminator.

### Discriminator

Discriminator menerima gambar asli maupun gambar hasil generasi kemudian menentukan apakah gambar tersebut termasuk kategori Real atau Fake.

Struktur Discriminator:

* Input Image (28×28)
* Convolution Layer
* LeakyReLU Activation
* Convolution Layer
* LeakyReLU Activation
* Fully Connected Layer
* Sigmoid Activation
* Output Probability (Real/Fake)

Discriminator berfungsi sebagai classifier biner yang memberikan umpan balik kepada gambar selama proses training.

## Hyperparameter Baseline

| Parameter        | Nilai                      |
| ---------------- | -------------------------- |
| Latent Dimension | 100                        |
| Batch Size       | 64                         |
| Epoch            | 50                         |
| Learning Rate    | 0.0002                     |
| Optimizer        | Adam                       |
| Loss Function    | Binary Cross Entropy (BCE) |

## Evaluasi

Kualitas gambar yang dihasilkan akan dievaluasi menggunakan Fréchet Inception Distance (FID). Metrik ini membandingkan distribusi feature antara gambar hasil generasi dan gambar asli menggunakan representasi feature dari jaringan Inception. Semakin kecil nilai FID, semakin mirip distribusi gambar hasil generasi terhadap distribusi gambar asli, yang menunjukkan kualitas image generation yang lebih baik.


```python
# import library
import os
import numpy as np
import matplotlib.pyplot as plt
import glob

from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader, random_split
from torchvision import transforms

from tqdm import tqdm
```

```python
# Load Gambar oil gas field
stadium_train = glob.glob(
    "version2/train/oil_gas_field/*.jpg"
)

stadium_test = glob.glob(
    "version2/test/stadium/*.jpg"
)

all_stadium = stadium_train + stadium_test

print("Total Images :", len(all_stadium))
```

```python
# transform
from torchvision import transforms

transform = transforms.Compose([
    transforms.Grayscale(),
    transforms.Resize((28,28)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])
```

```python
# custom dataset
from PIL import Image
from torch.utils.data import Dataset

class StadiumDataset(Dataset):

    def __init__(self, image_paths, transform=None):

        self.image_paths = image_paths
        self.transform = transform

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):

        image = Image.open(
            self.image_paths[idx]
        ).convert("L")

        if self.transform:
            image = self.transform(image)

        return image
```

```python
# dataset object
dataset = StadiumDataset(
    all_stadium,
    transform=transform
)

print("Dataset Size :", len(dataset))
```

```python
# split 80-10-10
from torch.utils.data import random_split

total_size = len(dataset)

train_size = int(0.8 * total_size)
val_size = int(0.1 * total_size)

test_size = (
    total_size
    - train_size
    - val_size
)

train_dataset, val_dataset, test_dataset = random_split(
    dataset,
    [train_size, val_size, test_size]
)

print("Train :", len(train_dataset))
print("Val   :", len(val_dataset))
print("Test  :", len(test_dataset))
```

```python
# Data Loader
from torch.utils.data import DataLoader

batch_size = 64

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False
)
```

```python
# visualisasi sample
import matplotlib.pyplot as plt

sample_batch = next(
    iter(train_loader)
)

plt.figure(figsize=(8,4))

for i in range(8):

    plt.subplot(2,4,i+1)

    img = sample_batch[i].squeeze()

    plt.imshow(
        img,
        cmap="gray"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()
```

## Data Preparation

Dataset Stadium berhasil dimuat dan dilakukan preprocessing berupa konversi grayscale, resizing menjadi 28×28 piksel, normalisasi ke rentang [-1,1], serta pembagian data menjadi training set, validation set, dan test set. Data training digunakan untuk melatih GAN, sedangkan data test digunakan untuk evaluasi menggunakan Fréchet Inception Distance (FID).

```python
# gambar (Baseline DCGAN)
import torch
import torch.nn as nn

class gambar(nn.Module):

    def __init__(self, latent_dim=100):

        super().__init__()

        self.model = nn.Sequential(

            nn.Linear(latent_dim, 128 * 7 * 7),

            nn.ReLU(True)
        )

        self.conv = nn.Sequential(

            nn.ConvTranspose2d(
                128,
                64,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.BatchNorm2d(64),

            nn.ReLU(True),

            nn.ConvTranspose2d(
                64,
                1,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.Tanh()
        )

    def forward(self, z):

        x = self.model(z)

        x = x.view(-1, 128, 7, 7)

        x = self.conv(x)

        return x
```

```python
# Discriminator (Baseline DCGAN)
class Discriminator(nn.Module):
    
    def __init__(self):

        super().__init__()

        self.model = nn.Sequential(

            nn.Conv2d(
                1,
                64,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.LeakyReLU(
                0.2,
                inplace=True
            ),

            nn.Conv2d(
                64,
                128,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.LeakyReLU(
                0.2,
                inplace=True
            ),

            nn.Flatten(),

            nn.Linear(
                128 * 7 * 7,
                1
            ),

            nn.Sigmoid()
        )

    def forward(self, x):

        return self.model(x)
```

```python
# Inisialisasi Model
device = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

latent_dim = 100

gambar = gambar(
    latent_dim
).to(device)

discriminator = Discriminator().to(device)

print(gambar)
print()
print(discriminator)
```

```python
# Lost Function and Optimizer
criterion = nn.BCELoss()

optimizer_G = torch.optim.Adam(
    gambar.parameters(),
    lr=0.0002,
    betas=(0.5, 0.999)
)

optimizer_D = torch.optim.Adam(
    discriminator.parameters(),
    lr=0.0002,
    betas=(0.5, 0.999)
)
```

## Training Baseline GAN

Model GAN dilatih menggunakan pendekatan adversarial learning. Pada setiap iterasi, Discriminator terlebih dahulu dilatih untuk membedakan gambar asli dan gambar hasil generasi. Selanjutnya gambar dilatih untuk menghasilkan gambar yang mampu menipu Discriminator.

Loss Function yang digunakan adalah Binary Cross Entropy (BCE), sedangkan optimizer yang digunakan adalah Adam dengan learning rate sebesar 0.0002.

```python
# Training Loop
epochs = 50

g_losses = []
d_losses = []

for epoch in range(epochs):

    g_loss_epoch = 0
    d_loss_epoch = 0

    for real_imgs in train_loader:

        real_imgs = real_imgs.to(device)

        batch_size = real_imgs.size(0)

        real_labels = torch.ones(
            batch_size,
            1
        ).to(device)

        fake_labels = torch.zeros(
            batch_size,
            1
        ).to(device)

        # ==================
        # Train Discriminator
        # ==================

        optimizer_D.zero_grad()

        outputs_real = discriminator(
            real_imgs
        )

        loss_real = criterion(
            outputs_real,
            real_labels
        )

        z = torch.randn(
            batch_size,
            latent_dim
        ).to(device)

        fake_imgs = gambar(z)

        outputs_fake = discriminator(
            fake_imgs.detach()
        )

        loss_fake = criterion(
            outputs_fake,
            fake_labels
        )

        d_loss = (
            loss_real
            + loss_fake
        ) / 2

        d_loss.backward()

        optimizer_D.step()

        # ==================
        # Train gambar
        # ==================

        optimizer_G.zero_grad()

        z = torch.randn(
            batch_size,
            latent_dim
        ).to(device)

        fake_imgs = gambar(z)

        outputs = discriminator(
            fake_imgs
        )

        g_loss = criterion(
            outputs,
            real_labels
        )

        g_loss.backward()

        optimizer_G.step()

        g_loss_epoch += g_loss.item()
        d_loss_epoch += d_loss.item()

    g_losses.append(
        g_loss_epoch / len(train_loader)
    )

    d_losses.append(
        d_loss_epoch / len(train_loader)
    )

    print(
        f"Epoch [{epoch+1}/{epochs}] "
        f"D Loss: {d_losses[-1]:.4f} "
        f"G Loss: {g_losses[-1]:.4f}"
    )
```

```python
# visualisasi training loss

plt.figure(figsize=(8,5))

plt.plot(
    g_losses,
    label="gambar Loss"
)

plt.plot(
    d_losses,
    label="Discriminator Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("GAN Training Loss")

plt.legend()

plt.show()
```

## Training Loss Analysis

Grafik di atas menunjukkan perkembangan gambar Loss dan Discriminator Loss selama proses training baseline GAN pada dataset Oil Gas Field.

Pada awal training (epoch 1–8), gambar Loss mengalami penurunan yang cukup signifikan dari nilai di atas 5 menjadi mendekati 1. Hal ini menunjukkan bahwa gambar dengan cepat mempelajari pola dasar pada gambar Oil Gas Field sehingga mampu menghasilkan gambar sintetis yang semakin mendekati distribusi data asli. Pada saat yang sama, Discriminator Loss meningkat secara bertahap hingga berada pada rentang sekitar 0.5–0.6.

Pada rentang epoch 8–30, gambar Loss dan Discriminator Loss menunjukkan pola yang relatif stabil. gambar Loss berada di sekitar nilai 0.9–1.1, sedangkan Discriminator Loss berada di sekitar 0.5–0.6. Kondisi ini mengindikasikan bahwa gambar dan Discriminator mulai mencapai keseimbangan sehingga proses adversarial learning berlangsung secara efektif.

Terdapat sedikit peningkatan gambar Loss pada sekitar epoch 34–37. Namun peningkatan ini tidak terlalu ekstrem dan segera kembali stabil pada epoch berikutnya. Fluktuasi ringan seperti ini merupakan fenomena yang umum terjadi selama training GAN karena gambar dan Discriminator saling berkompetisi untuk meningkatkan performanya masing-masing.

Pada akhir training (epoch 40–50), kedua loss cenderung konstan dan tidak menunjukkan osilasi yang berlebihan. Kondisi ini menunjukkan bahwa proses training relatif stabil dan tidak mengalami ketidakseimbangan yang signifikan antara gambar dan Discriminator.

Secara keseluruhan, hasil training menunjukkan bahwa baseline GAN pada dataset Oil Gas Field mampu mencapai proses adversarial learning yang lebih stabil dibandingkan kasus yang sering terjadi pada GAN. Oleh karena itu, kualitas akhir model akan dievaluasi menggunakan Fréchet Inception Distance (FID) untuk mengukur kesamaan distribusi feature antara gambar hasil generasi dan gambar asli.

# 3C - Improved GAN Architecture and Hyperparameter Tuning

## Architecture Improvements

### 1. Richer Feature Representation in gambar

Jumlah feature map pada gambar ditingkatkan sehingga model dapat mempelajari pola visual Oil Gas Field secara lebih mendalam.

Tujuan:

* Menghasilkan representasi gambar yang lebih kaya.
* Menangkap pola area industri yang lebih detail.
* Meningkatkan kualitas feature yang dihasilkan pada proses generasi.

### 2. Dropout in Discriminator

Dropout ditambahkan pada Discriminator untuk meningkatkan kemampuan generalisasi.

Tujuan:

* Mengurangi ketergantungan terhadap pola tertentu pada data training.
* Memberikan feedback yang lebih beragam kepada gambar.
* Membantu menghasilkan gambar yang lebih bervariasi.

### 3. DCGAN Weight Initialization

Weight initialization khusus DCGAN digunakan untuk menghasilkan distribusi bobot awal yang lebih optimal.

Tujuan:

* Mempercepat proses pembelajaran.
* Membantu gambar menghasilkan feature yang lebih representatif sejak awal training.

Dengan representasi feature yang lebih kaya, proses optimasi yang lebih halus, dan kemampuan generalisasi yang lebih baik, model diharapkan menghasilkan nilai FID yang lebih rendah dibandingkan baseline GAN sebesar 135.5490.

Penurunan nilai FID akan menunjukkan bahwa gambar mampu menghasilkan gambar Oil Gas Field yang lebih realistis dan memiliki distribusi feature yang semakin mendekati gambar asli.


```python
# improved gambar oil gas
class Improvedgambar(nn.Module):
    
    def __init__(self, latent_dim=100):

        super().__init__()

        self.fc = nn.Sequential(

            nn.Linear(
                latent_dim,
                256 * 7 * 7
            ),

            nn.BatchNorm1d(
                256 * 7 * 7
            ),

            nn.ReLU(True)
        )

        self.conv = nn.Sequential(

            nn.ConvTranspose2d(
                256,
                128,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.BatchNorm2d(128),

            nn.ReLU(True),

            nn.ConvTranspose2d(
                128,
                64,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.BatchNorm2d(64),

            nn.ReLU(True),

            nn.Conv2d(
                64,
                1,
                kernel_size=3,
                stride=1,
                padding=1
            ),

            nn.Tanh()
        )

    def forward(self, z):

        x = self.fc(z)

        x = x.view(
            -1,
            256,
            7,
            7
        )

        x = self.conv(x)

        return x
```

```python
# improved discriminator
class ImprovedDiscriminator(nn.Module):
    
    def __init__(self):

        super().__init__()

        self.model = nn.Sequential(

            nn.Conv2d(
                1,
                64,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.LeakyReLU(
                0.2,
                inplace=True
            ),

            nn.Dropout2d(
                0.3
            ),

            nn.Conv2d(
                64,
                128,
                kernel_size=4,
                stride=2,
                padding=1
            ),

            nn.BatchNorm2d(
                128
            ),

            nn.LeakyReLU(
                0.2,
                inplace=True
            ),

            nn.Dropout2d(
                0.3
            ),

            nn.Flatten(),

            nn.Linear(
                128 * 7 * 7,
                1
            ),

            nn.Sigmoid()
        )

    def forward(self, x):

        return self.model(x)
```

```python
# weight initialization
def weights_init(m):
    
    classname = m.__class__.__name__

    if classname.find("Conv") != -1:

        nn.init.normal_(
            m.weight.data,
            0.0,
            0.02
        )

    elif classname.find("BatchNorm") != -1:

        nn.init.normal_(
            m.weight.data,
            1.0,
            0.02
        )

        nn.init.constant_(
            m.bias.data,
            0
        )
```

```python
# initialize model
latent_dim = 100

gambar_imp = Improvedgambar(
    latent_dim
).to(device)

discriminator_imp = ImprovedDiscriminator().to(device)

gambar_imp.apply(
    weights_init
)

discriminator_imp.apply(
    weights_init
)
```

```python
# optimizer
criterion = nn.BCELoss()

optimizer_G_imp = torch.optim.Adam(
    gambar_imp.parameters(),
    lr=0.0001,
    betas=(0.5, 0.999)
)

optimizer_D_imp = torch.optim.Adam(
    discriminator_imp.parameters(),
    lr=0.0001,
    betas=(0.5, 0.999)
)
```

```python
epochs = 75

g_losses_imp = []
d_losses_imp = []

for epoch in range(epochs):

    g_loss_epoch = 0
    d_loss_epoch = 0

    for real_imgs in train_loader:

        real_imgs = real_imgs.to(device)

        batch_size = real_imgs.size(0)

        real_labels = torch.ones(
            batch_size,
            1,
            device=device
        )

        fake_labels = torch.zeros(
            batch_size,
            1,
            device=device
        )

        # ==================================
        # Train Improved Discriminator
        # ==================================

        optimizer_D_imp.zero_grad()

        outputs_real = discriminator_imp(
            real_imgs
        )

        loss_real = criterion(
            outputs_real,
            real_labels
        )

        z = torch.randn(
            batch_size,
            latent_dim,
            device=device
        )

        fake_imgs = gambar_imp(z)

        outputs_fake = discriminator_imp(
            fake_imgs.detach()
        )

        loss_fake = criterion(
            outputs_fake,
            fake_labels
        )

        d_loss = (
            loss_real + loss_fake
        ) / 2

        d_loss.backward()

        optimizer_D_imp.step()

        # ==================================
        # Train Improved gambar
        # ==================================

        optimizer_G_imp.zero_grad()

        z = torch.randn(
            batch_size,
            latent_dim,
            device=device
        )

        fake_imgs = gambar_imp(z)

        outputs = discriminator_imp(
            fake_imgs
        )

        g_loss = criterion(
            outputs,
            real_labels
        )

        g_loss.backward()

        optimizer_G_imp.step()

        g_loss_epoch += g_loss.item()
        d_loss_epoch += d_loss.item()

    avg_g_loss = (
        g_loss_epoch
        / len(train_loader)
    )

    avg_d_loss = (
        d_loss_epoch
        / len(train_loader)
    )

    g_losses_imp.append(
        avg_g_loss
    )

    d_losses_imp.append(
        avg_d_loss
    )

    print(
        f"Epoch [{epoch+1}/{epochs}] "
        f"D Loss: {avg_d_loss:.4f} "
        f"G Loss: {avg_g_loss:.4f}"
    )
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    g_losses_imp,
    label="Improved gambar Loss"
)

plt.plot(
    d_losses_imp,
    label="Improved Discriminator Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Improved GAN Training Loss")

plt.legend()

plt.show()
```
